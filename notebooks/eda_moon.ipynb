```python
# 지원 1단계 step 거친 user_uuid 
step_1 = applying_df['URL'].str.contains('api/jobs/id/apply/step1')
step_1_users = applying_df.loc[step_1, 'user_uuid'].nunique()
step_1_users
```

```python
# 지원 2단계 step 거친 user_uuid 
step_2 = applying_df['URL'].str.contains('api/jobs/id/apply/step2')
step_2_users = applying_df.loc[step_2, 'user_uuid'].nunique()
step_2_users
```

```python
# 지원 3단계 step 거친 user_uuid 
step_3 = applying_df['URL'].str.contains('api/jobs/id/apply/step3')
step_3_users = applying_df.loc[step_3, 'user_uuid'].nunique()
step_3_users
```

```python
# 지원 4단계 step 거친 user_uuid 
step_4 = applying_df['URL'].str.contains('api/jobs/id/apply/step4')
step_4_users = applying_df.loc[step_4, 'user_uuid'].nunique()
step_4_users
```

```python
# 도달율 구하기 
reach_rates = {
    "apply_step1": step_1_users / applying_users,
    "apply_step2": step_2_users / applying_users,
    "apply_step3": step_3_users / applying_users,
    "apply_step4": step_4_users / applying_users,
}

{ k: f"{v*100:.2f}%" for k, v in reach_rates.items() }
```

```python
# 순차 퍼널 
step1_set = set(
    applying_df.loc[applying_df['URL'].str.contains('api/jobs/id/apply/step1'), 'user_uuid']
)

step2_set = set(
    applying_df.loc[applying_df['URL'].str.contains('api/jobs/id/apply/step2'), 'user_uuid']
)

step3_set = set(
    applying_df.loc[applying_df['URL'].str.contains('api/jobs/id/apply/step3'), 'user_uuid']
)

step4_set = set(
    applying_df.loc[applying_df['URL'].str.contains('api/jobs/id/apply/step4'), 'user_uuid']
)

####
strict_step2 = step1_set & step2_set
strict_step3 = strict_step2 & step3_set
strict_step4 = strict_step3 & step4_set

strict_counts = {
    "step1": len(step1_set),
    "step2": len(strict_step2),
    "step3": len(strict_step3),
    "step4": len(strict_step4),
}
strict_counts
```

```python
rates = {
    "step1_to_step2": {
        "conversion_rate": len(strict_step2) / len(step1_set),
        "drop_rate": 1 - (len(strict_step2) / len(step1_set)),
    },
    "step2_to_step3": {
        "conversion_rate": len(strict_step3) / len(strict_step2),
        "drop_rate": 1 - (len(strict_step3) / len(strict_step2)),
    },
    "step3_to_step4": {
        "conversion_rate": len(strict_step4) / len(strict_step3),
        "drop_rate": 1 - (len(strict_step4) / len(strict_step3)),
    },
}

pretty_rates = {
    k: {
        "conversion_rate": f"{v['conversion_rate']*100:.2f}%",
        "drop_rate": f"{v['drop_rate']*100:.2f}%",
    }
    for k, v in rates.items()
}

pretty_rates
```
