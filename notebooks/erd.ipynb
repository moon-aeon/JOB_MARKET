```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine
import yaml
```

```python
with open("../config-local.yaml", "r", encoding="utf-8") as f:
    config = yaml.safe_load(f)

mysql_config = config["db"]
engine = create_engine(
    f"mysql+pymysql://{mysql_config['user']}:{mysql_config['password']}@{mysql_config['host']}:{mysql_config['port']}/{mysql_config['name']}?charset=utf8mb4"
)

pd.read_sql("SELECT 1 AS ok;", engine)
```

```python
app_raw_df = pd.read_sql_table("Application", con=engine)
app_df = app_raw_df.copy()
app_df.head()
```
